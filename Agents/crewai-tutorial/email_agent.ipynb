```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from crewai import  LLM
llm = LLM(
    model="gemini/gemini-2.0-flash",
    temperature=0.1
)
llm.call("Who invented transcendental meditation.")
```

```python
from crewai import Agent, Task, Crew

email_assistant = Agent(
    role="Email Assistant Agent",
    goal="Improve emails and make them sound professional and clear",
    backstory="A highly experienced communicatiob expert skilled in professional email writtin",
    verbose=True,
    llm=llm
    
)

original_email = """
hey team, just want to tell u that the demo is kins of ready, but there's still stuff left.
Maybe we can show what we have and say rest is WIP.
Let me know what u think. thanks
"""

email_task = Task(
    description=f"""Take the following rough email and rewrite it into a professional and polished 
    version.
    Exapand abbrevations:
    '''{original_email}''' """,
    agent=email_assistant,
    expected_output="A professional written email with proper formatting and content.",
)

crew = Crew(
    agents=[email_assistant],
    tasks=[email_task],
    verbose=True
)

result = crew.kickoff()
print(result)
```
